```python
# Take a look at what's in our newly downloaded and unzipped folder

xml_files = glob.glob("*.xml")
xml_files.sort()
xml_files
```

**Load the Raw Dataset**

```python
import pandas as pd

pred_df = pd.read_excel('Herbal coba.xlsx', sheet_name = 'Sheet1')
pred_df
```

**Fingerprints Generation**

```python
# Convert our file to smi format

clean_fp_df = pd.concat( [pred_df['SMILES'],pred_df['Active Compound']], axis=1 )
clean_fp_df.to_csv('clean_df.smi', sep='\t', index=False, header=False)
clean_fp_df.head()
```

```python
# Fingerprints making process

for i in range(len(xml_files)):

  padeldescriptor(mol_dir='clean_df.smi', d_file=f'{xml_files[i][:-4]}.csv',
                  descriptortypes= xml_files[i],retainorder=True,
                  fingerprints=True,d_2d=False,
                  d_3d=False)
  fp_name = f'{xml_files[i][:-4]}'
  fp = pd.read_csv(f'{fp_name}.csv')
  fp = pd.concat([clean_fp_df, fp.drop('Name', axis=1)], axis=1)
  fp.to_csv(f'{fp_name}.csv', index=None)
  print (f'{xml_files[i][:-4]} done')
```

# **Loading the Saved Model and Predicting**

```python
import joblib
import pandas as pd

# Load your trained model + label encoder

saved = joblib.load("bayes_search_lgbm_pipeline.joblib")
model = saved["model"]
label_encoder = saved["label_encoder"]
```

```python
# Load the Dataset and Dropping the unnecessary columns

merged_df = pd.read_excel('merged_uji_fp.xlsx')
X_test = merged_df.drop(columns=["Active Compound", "SMILES"], errors='ignore')
X_test
```

```python
# Make predictions

predictions = model.predict(X_test)
predicted_labels = label_encoder.inverse_transform(predictions)
```

```python
# Save results

clean_fp_df["Predicted_Class"] = predicted_labels
clean_fp_df.to_csv("test_with_predictions.csv", index=False)

print("✅ Done! Predictions saved to 'test_with_predictions.csv'")
```
